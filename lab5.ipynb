```python
import pandas as pd

df=pd.read_csv("Student_Performance.csv")
Q1=df["Hours Studied"].quantile(0.25)
Q3=df["Hours Studied"].quantile(0.75)

IQR=Q3-Q1

lower = Q1-1.5*IQR
upper = Q3+1.5*IQR

outliers_iqr=df[
   (df["Hours Studied"]<lower)|
   (df["Hours Studied"]>upper)
    ]

print(f'IQR bounds:[{lower:.2f},{upper:.2f}]')
print('Outliers found(IQR):',len(outliers_iqr))

df['Z_Score'] =(
    df['Hours Studied']- df['Hours Studied'].mean()
)/ df['Hours Studied'].std()

outliers_z = df[df["Z_Score"].abs() > 3]
print('Outliers found(Z-score >3):',len(outliers_z))


#capping
df["Hours_Studied_Capped"] = df["Hours Studied"].clip(lower = lower, upper=upper)
print('\nBefore vs after capping (max value):')
print('Before:',df['Hours Studied'].max())
print('After:',df['Hours_Studied_Capped'].max())

#mean = df["Hours Studied"].mean()
#std = df["Hours Studied"].std()    
```
