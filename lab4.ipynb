```python

import pandas as pd
import numpy as np

# Create a sample dataset
data = {
    'Name': ['Ravi', 'Priya', 'Arun', 'Sneha', 'Rahul'],
    'Age': [20, 21, np.nan, 23, 22],
    'Marks': [80, 90, 70, np.nan, 60],
    'City': ['Bangalore', 'Mysore', 'Bangalore', np.nan, 'Mysore'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# 1. Handle missing values using Mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

# 2. Handle missing values using Median
df['Marks'] = df['Marks'].fillna(df['Marks'].median())

# 3. Handle missing values using Forward Fill
df['City'] = df['City'].ffill()

print("\nDataset After Handling Missing Values:")
print(df)

# 4. Encode categorical variables using One-Hot Encoding
df = pd.get_dummies(
    df,
    columns=['Gender', 'City'],
    dtype=int
)

print("\nFinal Preprocessed Dataset:")
print(df)

# Check for remaining missing values
print("\nMissing Values in Each Column:")
print(df.isnull().sum())
```
